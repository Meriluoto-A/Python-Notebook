```python
# GOAL OF THIS CODE
# ---------------------------------------------------------
# This script practices data analysis and visualization # using Seaborn. It calculates:
# - total yearly ice cream sales
# - average (mean) monthly sales
# - highest and lowest monthly sales
# - which months had the most and least sales
# It also creates two Seaborn charts:
# 1) Bar chart of monthly sales
# 2) Line chart showing the sales trend
# ---------------------------------------------------------

import seaborn as sns
import matplotlib.pyplot as plt

# ---------------------------------------------------------
# DATA: Ice cream sales per month (kEUR)
# ---------------------------------------------------------

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

sales = [5, 4, 4, 6, 10, 15, 20, 21, 15, 10, 5, 5]   # in kEUR

# ---------------------------------------------------------
# BASIC CALCULATIONS
# ---------------------------------------------------------

total_sales = sum(sales)                     # total yearly sales
average_sales = sum(sales) / len(sales)      # mean
max_sales = max(sales)                       # highest value
max_month = months[sales.index(max_sales)]   # month of highest value
min_sales = min(sales)                       # lowest value
min_month = months[sales.index(min_sales)]   # month of lowest value

# Print results
print("Total yearly sales (kEUR):", total_sales)
print("Average monthly sales (kEUR):", average_sales)
print("Highest sales:", max_sales, "kEUR in", max_month)
print("Lowest sales:", min_sales, "kEUR in", min_month)

# ---------------------------------------------------------
# SEABORN BAR CHART
# ---------------------------------------------------------

plt.figure(figsize=(12, 6))
sns.barplot(x=months, y=sales, color="green")

plt.title("Ice Cream Sales per Month (kEUR) — Bar Chart")
plt.xlabel("Month")
plt.ylabel("Sales (kEUR)")
plt.grid(True)
plt.show()
```
