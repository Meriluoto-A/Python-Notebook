```python
# ==========================================
# STRESS MANAGEMENT MATRIX
# ==========================================
#
# The Stress Management Matrix helps organise
# concerns according to:
#
# 1. Stress Level (0–10)
# 2. Ability to Influence (0–10)
#
# The purpose is to focus energy on issues
# that can be influenced while reducing
# attention on issues that cannot be controlled.
#
# Quadrants:
#
# ACT      = High stress, High influence
# ACCEPT   = High stress, Low influence
# MONITOR  = Low stress, High influence
# IGNORE   = Low stress, Low influence
#
# ==========================================

import pandas as pd
import matplotlib.pyplot as plt

# ==========================================
# INPUT DATA
# ==========================================

data = {
    "Issue": [
        "Job Interview",
        "World Economy",
        "Essay Feedback",
        "Electricity Price",
        "Learning Python",
        "Traffic"
    ],
    "Stress": [8, 7, 6, 5, 4, 2],
    "Influence": [7, 1, 9, 2, 8, 1]
}

df = pd.DataFrame(data)

# ==========================================
# ASSIGN QUADRANTS
# ==========================================

def assign_quadrant(row):

    if row["Stress"] >= 5 and row["Influence"] >= 5:
        return "Act"

    elif row["Stress"] >= 5 and row["Influence"] < 5:
        return "Accept"

    elif row["Stress"] < 5 and row["Influence"] >= 5:
        return "Monitor"

    else:
        return "Ignore"


df["Quadrant"] = df.apply(assign_quadrant, axis=1)

# ==========================================
# MATRIX VISUALISATION
# ==========================================

fig, ax = plt.subplots(figsize=(10, 8))

# Background colours

# ACCEPT
ax.axvspan(
    0, 5,
    ymin=0.5,
    ymax=1,
    color="#FFF3CD",
    alpha=0.6
)

# ACT
ax.axvspan(
    5, 10,
    ymin=0.5,
    ymax=1,
    color="#F8D7DA",
    alpha=0.6
)

# IGNORE
ax.axvspan(
    0, 5,
    ymin=0,
    ymax=0.5,
    color="#E2E3E5",
    alpha=0.6
)

# MONITOR
ax.axvspan(
    5, 10,
    ymin=0,
    ymax=0.5,
    color="#D1ECF1",
    alpha=0.6
)

# Quadrant lines

ax.axhline(
    5,
    color="black",
    linestyle="--",
    linewidth=1
)

ax.axvline(
    5,
    color="black",
    linestyle="--",
    linewidth=1
)

# Bubble colours

colour_map = {
    "Act": "#D9534F",       # Red
    "Accept": "#F0AD4E",    # Yellow / Orange
    "Monitor": "#5BC0DE",   # Blue
    "Ignore": "#A9A9A9"     # Grey
}

# Plot bubbles

for _, row in df.iterrows():

    ax.scatter(
        row["Influence"],
        row["Stress"],
        s=700,
        color=colour_map[row["Quadrant"]],
        edgecolors="black",
        alpha=0.85
    )

    ax.text(
        row["Influence"] + 0.15,
        row["Stress"] + 0.15,
        row["Issue"],
        fontsize=11
    )

# Quadrant Labels

ax.text(
    7.2,
    8.6,
    "ACT",
    fontsize=18,
    fontweight="bold"
)

ax.text(
    1.0,
    8.6,
    "ACCEPT",
    fontsize=18,
    fontweight="bold"
)

ax.text(
    7.0,
    1.4,
    "MONITOR",
    fontsize=18,
    fontweight="bold"
)

ax.text(
    1.0,
    1.4,
    "IGNORE",
    fontsize=18,
    fontweight="bold"
)

# Formatting

ax.set_xlim(0, 10)
ax.set_ylim(0, 10)

ax.set_xlabel(
    "Ability to Influence (0-10)",
    fontsize=12
)

ax.set_ylabel(
    "Stress Level (0-10)",
    fontsize=12
)

ax.set_title(
    "Stress Management Matrix",
    fontsize=16,
    fontweight="bold"
)

plt.tight_layout()
plt.show()

# ==========================================
# RESULTS TABLE
# ==========================================

print("\nResults\n")

results = df.sort_values(
    by=["Quadrant", "Stress"],
    ascending=[True, False]
)

results = df.sort_values(
    by=["Quadrant", "Stress"],
    ascending=[True, False]
)

display(results)
```
