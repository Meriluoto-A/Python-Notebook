```python
#There are several Datatypes in Python. Here is explanation which they are

# INTEGER
print("INTEGER:")
print("Integer is a whole number without decimals.")
age = 30
print("Example:", age)
print()

# FLOAT
print("FLOAT:")
print("Float is a number with decimals.")
temperature = 19.0
print("Example:", temperature)
print()

# STRING
print("STRING:")
print("String is text inside quotes.")
name = "Anne"
print("Example:", name)
print()

# BOOLEAN
print("BOOLEAN:")
print("Boolean is either True or False.")
is_ready = True
print("Example:", is_ready)
print()

# LIST
print("LIST:")
print("List is an ordered, changeable collection of items.")
fruits = ["apple", "banana", "cherry"]
print("Example:", fruits)
print()

# TUPLE
print("TUPLE:")
print("Tuple is an ordered, unchangeable collection of items.")
coordinates = (60.17, 24.94)
print("Example:", coordinates)
print()

# DICTIONARY
print("DICTIONARY:")
print("Dictionary stores key–value pairs.")
person = {"name": "Anne", "age": 30}
print("Example:", person)
print()

# ASKING WHICH TYPE
print("ASKING WHICH TYPE:")
print("Use type() to check the data type of a value.")
print("Example: type(age) =", type(age))
print("Example: type(fruits) =", type(fruits))
print("Example: type(8.7) =", type(8.7))
print("Example: type(5) =", type(5))
      
```
