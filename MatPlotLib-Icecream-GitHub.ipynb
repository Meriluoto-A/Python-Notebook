```python
# ---------------------------------------------------------
# GOAL OF THIS CODE
# ---------------------------------------------------------
# This script practices basic data analysis with Python.
# It calculates:
# - average (mean) ice cream sales
# - highest and lowest monthly sales
# - which months had the most and least sales
# It also creates a simple Matplotlib figure showing ice cream sales for each month.
# ---------------------------------------------------------

import matplotlib.pyplot as plt

# ---------------------------------------------------------
# DATA: Ice cream sales per month (kEUR)
# ---------------------------------------------------------

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

sales = [5, 4, 4, 6, 10, 15, 20, 21, 15, 10, 5, 5]   # in kEUR

# ---------------------------------------------------------
# BASIC CALCULATIONS
# ---------------------------------------------------------

# Average (mean) sales: sum of all sales divided by number of months
average_sales = sum(sales) / len(sales)

# Highest sales value
max_sales = max(sales)

# Month with highest sales (find index of max value)
max_month = months[sales.index(max_sales)]

# Lowest sales value
min_sales = min(sales)

# Month with lowest sales (find index of min value)
min_month = months[sales.index(min_sales)]

# Print results
print("Average sales (kEUR):", average_sales)
print("Highest sales:", max_sales, "kEUR in", max_month)
print("Lowest sales:", min_sales, "kEUR in", min_month)

# ---------------------------------------------------------
# MATPLOTLIB FIGURE
# ---------------------------------------------------------

plt.figure(figsize=(5,2))  # set figure size

# Line plot of monthly sales
plt.plot(months, sales, marker="o", linestyle="-", color="green")

# Add title and labels
plt.title("Ice Cream Sales per Month (kEUR)")
plt.xlabel("Month")
plt.ylabel("Sales (kEUR)")

# Add grid for readability
plt.grid(True)

# Show the figure
plt.show()

```
