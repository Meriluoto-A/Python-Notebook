```python
#  SUDOKU AS A LIST-------------------

# A Sudoku grid must be a 2D list (list of lists)
Sudoku = [
    [5, 3, 0, 0, 7, 0, 0, 0, 0],
    [6, 0, 0, 1, 9, 5, 0, 0, 0],
    [0, 9, 8, 0, 0, 0, 0, 6, 0],
    [8, 0, 0, 0, 6, 0, 0, 0, 3],
    [4, 0, 0, 8, 0, 3, 0, 0, 1],
    [7, 0, 0, 0, 2, 0, 0, 0, 6],
    [0, 6, 0, 0, 0, 0, 2, 8, 0],
    [0, 0, 0, 4, 1, 9, 0, 0, 5],
    [0, 0, 0, 0, 8, 0, 0, 7, 9]
]

def print_grid(Sudoku):
    # Loop through each row in the grid
    for row in Sudoku:
        # Loop through each square in the row
        for square in row:
            # If the square has a number, print it
            if square > 0:
                print(f" {square}", end="")
            # If the square is 0, print an underscore
            else:
                print(" _", end="")
        print()  # Move to next line after each row

# Print the grid
print_grid(Sudoku)

print() # empty row
print() # empty row


# ---------------------------------------------------------
# Generate a 9x9 Sudoku-like grid with random numbers
# Half of the squares will be empty (0)
# ---------------------------------------------------------

import random

def generate_random_sudoku():
    sudoku = []

    for _ in range(9):  # 9 rows
        row = []
        for _ in range(9):  # 9 columns
            # 50% chance the square is empty
            if random.random() < 0.5:
                row.append(0)
            else:
                row.append(random.randint(1, 9))  # random number 1–9
        sudoku.append(row)

    return sudoku

# ---------------------------------------------------------
# Print the grid nicely
# ---------------------------------------------------------
def print_grid(Sudoku):
    for row in Sudoku:
        for square in row:
            if square > 0:
                print(f" {square}", end="")
            else:
                print(" _", end="")
        print()  # new line after each row

# ---------------------------------------------------------
# Create and print a new random Sudoku
# ---------------------------------------------------------
Sudoku = generate_random_sudoku()
print_grid(Sudoku)

print()  # empty line
print()
```
