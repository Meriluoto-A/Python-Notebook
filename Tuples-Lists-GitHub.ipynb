```python
# Tuples = () = fixed
# Lists = [] = changeable

# ---------------------------------------------------------
# LISTS vs TUPLES — Coffee Examples
# ---------------------------------------------------------

# A LIST: mutable (changeable), uses square brackets []
coffee_list = ["espresso", "latte", "cappuccino", "mocha"]

print("Original coffee list:", coffee_list)

# Lists can be changed
coffee_list.append("flat white")   # add new item
coffee_list[1] = "iced latte"      # modify an item
print("Modified coffee list:", coffee_list)


# ---------------------------------------------------------
# A TUPLE: immutable (cannot be changed), uses parentheses ()
coffee_tuple = ("espresso", "latte", "cappuccino", "mocha")

print("\nOriginal coffee tuple:", coffee_tuple)

# Trying to modify a tuple will cause an error:
# coffee_tuple[1] = "iced latte"   # This would crash the program

# Tuples are good for fixed data that should not change
sizes_tuple = ("small", "medium", "large")
print("Coffee sizes (tuple):", sizes_tuple)


# ---------------------------------------------------------
# Using both together
# Example: list of coffees + tuple of sizes
print("\nCoffee menu:")
for coffee in coffee_list:
    for size in sizes_tuple:
        print(f"{size} {coffee}")
```
