```python
# -----------------------------------------
# Coffee list stored inside brackets []
# -----------------------------------------

x = ["Espresso", "Cappuccino", "Latte", "Macchiato", "Lungo", "Flat White"]

# -----------------------------------------
# Print specific items using indexing
# Indexing starts at 0
# -----------------------------------------

print(x[4])        # Prints item at index 4 → Lungo
print(x[0:5])      # Prints items from index 0 up to (but not including) 5
print(x[3:4])      # Prints item at index 3 as a list → ['Macchiato']

# -----------------------------------------
# Check the type of x
# -----------------------------------------

print(type(x))     # <class 'list'>

# -----------------------------------------
# Adding new coffees to the list
# append() adds one item to the end
# -----------------------------------------

x.append("Ristretto")   # Add new coffee
x.append("Mocha")       # Add another coffee

# -----------------------------------------
# Print updated list
# -----------------------------------------

print(x)
```
